```python
Input:
[1, 2, 2, 3, 1, 1, 4]

Output:
{
    1: 3,
    2: 2,
    3: 1,
    4: 1
}
```

```python
def aa(a):
    dic = {}

    for i in a:
        if i in dic:
            dic[i] = dic[i] + 1
        else:
            dic[i] = 1

    return dic
```

```python
aa([1, 2, 2, 3, 1, 1, 4])
```

```python
def mn(a):
    i1 = a[0]
    for i in a:
        if i > i1:
            i1=i

    return i1
```
